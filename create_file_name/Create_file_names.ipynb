# Rounding

```python
"""
Usfeul function to formating the values of variables.
"""

def rounded_number(_number, _significant_digits):
    """
    Taking the value to rounded and and takes the amount of significant digits you specify.
    """
    return round(_number, _significant_digits - int(math.floor(math.log10(abs(_number)))) - 1)

# checking

number = 1234.23
significant_digits = 3

number_after_rounded = rounded_number(number, significant_digits)
print(number_after_rounded)
```

# Fixing variables (mass, time, sigma)

```python
"""
Denotion:
    diff: different
    coff: coefficient
    val:  value
"""
def Mass_power(_val):
    """
    _val: mass value of the Galaxy [GeV]
    """
    return math.log10(_val)


def Calculate_val(_coff, _power):
    """
    _coff: coefficient
    _pow:  power
    """
    return (_coff * 10**_power)

# powers and coefficients, which set mass of galactic
list_set_pow_mass = [9.0,10.0,11.0,12.0]
diff_power_mass = len(list_set_pow_mass)

list_coff_mass = [1.0, 2.0, 5.0]
diff_coff_mass = len(list_coff_mass)

# values of masses - for every coefficients and every power we calculate mass of galaxy
list_mass = [] # [GeV]
list_power_mass = []

for power in list_set_pow_mass:
    for coff in list_coff_mass:
        # value of galactic mass
        val_mass = Calculate_val(coff, power)
        list_mass.append(val_mass)
        """
         only the power of mass (we colect that because we 
         have to set data in specific manner)
        """
        pow_mass = Mass_power(val_mass)
        list_power_mass.append(pow_mass)

# remove last few elements from lists
for i in (0, diff_coff_mass-1):
    list_mass.pop()
    list_power_mass.pop()
    
print("list_power_mass")
print(list_power_mass)
# print(list_mass)

"----------------------------------------------------------------------------------------------------------------"

# powers and coefficients, which set cross section
list_pow_sigma = [-1.0, 0.0, 1.0, 2.0, 3.0]
diff_pow_sigma = len(list_pow_sigma)

list_coff_sigma = [1.0, 2.0, 5.0]
diff_coff_sigma = len(list_coff_sigma)

# values of cross section - for every coefficient and every power.
list_sigma = [] # [cm^2/g]

for power in list_pow_sigma:
    for coff in list_coff_sigma:
        
        val_cross = Calculate_val(coff, power)
        list_sigma.append(val_cross)
        
# remove last few elements from lists
for i in (0, diff_coff_sigma-1):
    list_sigma.pop()
        
print()
print("list_sigma")
print(list_sigma)

list_time = [10.0] # [Gyr]
```
